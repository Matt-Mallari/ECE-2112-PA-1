## 1. Word Rotation Function

```python
def rotate_word(text: str) -> str:
    if not isinstance(text, str) or not text.strip():
        raise ValueError("Argument must be a non-empty string.")
    if len(text) <= 1:
        return text
    return text[1:] + text[0]
    
```

#### *Function 1 Sample Output:*

```python
print(rotate_word("python"))
print(rotate_word("logic"))
print(rotate_word("Code"))   
print(rotate_word("A"))      
```

## 2. Username Builder Function

```python
def make_username(first_name: str, last_name: str) -> str:
    clean_first = first_name.lower().replace(" ", "")
    clean_last = last_name.lower().replace(" ", "")
    return clean_first + "." + clean_last
```

#### *Function 2 Sample Output:*

```python
print(make_username("Ada", "Lovelace"))
print(make_username("Alan", "Turing"))
print(make_username("Ana Maria", "De Leon"))
```

## 3. Bookend Swap Function

```python
def swap_bookends(items: list) -> list:
    if len(items) < 2:
        raise ValueError("List must contain at least two elements.")
    first, *middle, last = items
    return [last, *middle, first]
```

#### *Function 3 Sample Output:*

```python
print(swap_bookends([1, 2, 3, 4, 5, 6]))
print(swap_bookends(["red", "green", "blue"]))
print(swap_bookends([8, 3]))
```
